# Ex1 - Filtering and Sorting Data



### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv) and assign it to a variable called chipo.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")
chipo
```

### Step 3. Name of the max valued product

```python
chipo['item_price_'] = chipo['item_price'].apply(lambda x: float(x[1:]))
# Otra forma: 
# chipo['item_price_'] = pd.to_numeric(chipo['item_price'].astype(str).str.replace("$", ""))
chipo['unit_price'] = chipo['item_price_'] / chipo['quantity']
# chipo.loc[chipo['unit_price'].idxmax(), 'item_name']
chipo.loc[(chipo['unit_price'] == chipo['unit_price'].max()), 'item_name'].drop_duplicates().values
```

### Step 4. How many products cost more than $10.00?

```python
chipo.loc[(chipo['unit_price'] > 10), 'item_name'].nunique()
```

### Step 4.1: Y cuántos pedidos se han hecho con un producto de más de 10$? Es lo mismo?

```python
print(chipo.loc[(chipo['unit_price'] > 10), 'order_id'].nunique())
print('No es lo mismo.')
```

### Step 4.2: Y cuántos pedidos se han hecho de más de 10$? Es lo mismo?

```python
chipo_costes_pedidos = chipo.groupby('order_id')['item_price_'].sum()
# chipo_pedidos_m10 = chipo_costes_pedidos[(chipo_costes_pedidos['item_price_'] > 15)]
print(len(chipo_costes_pedidos[chipo_costes_pedidos > 10]))
print(chipo_costes_pedidos[chipo_costes_pedidos > 10].shape[0])
print('No es lo mismo.')
```

### Step 4.3: Y en cuántos pedidos se ha pagado más de 10$ por un mismo producto? Es lo mismo?

```python
print(chipo.loc[(chipo['item_price_'] > 10), 'order_id'].nunique())
print('No es lo mismo.')
```

### Step 5. What is the price of each item and name it unit_price. Get only item_name and unit_price

```python
# Hecho en paso 3: 
# chipo['unit_price'] = chipo['item_price_'] / chipo['quantity']

# Paso 5: 
# chipo[['item_name','unit_price']]
chipo_precios_productos = chipo.drop_duplicates(subset=['item_name', 'unit_price'])[['item_name', 'unit_price']]
chipo_precios_productos
```

### Step 6. Sort by the name of the item

```python
chipo_precios_productos.sort_values('item_name')
```

### Step 7. What was the quantity of the most expensive item ordered? 2 ways

```python
# 1. 
print(chipo.loc[(chipo['item_price_'] == chipo['item_price_'].max()), 'quantity'].values)
print()
# 2. 
print(chipo.loc[chipo['item_price_'].idxmax(), 'quantity'])
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
print(len(chipo[(chipo['item_name'] == 'Veggie Salad Bowl')]))
# Para obtener el nº total de unidades de este producto que se pidió (puede ser que en algunos pedidos se pida más de una)
print(chipo.loc[chipo['item_name'] == "Veggie Salad Bowl", "quantity"].sum())
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
len(chipo[(chipo['item_name'] == 'Canned Soda') & (chipo['quantity'] > 1.1)])
```
