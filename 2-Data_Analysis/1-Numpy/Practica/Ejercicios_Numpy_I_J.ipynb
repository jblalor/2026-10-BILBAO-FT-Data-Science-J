## Ejercicio 8
Concatena los siguientes arrays verticalmente

```Python
a = np.arange(10).reshape(2,-1)
b = np.repeat(1, 10).reshape(2,-1)
```

````python
a = np.arange(10).reshape(2,-1)
b = np.repeat(1, 10).reshape(2,-1)
c1 = np.concatenate([a,b], axis=0)
c2 = np.vstack([a, b])
print(a)
print()
print(b)
print()
print(c1)
print()
print(c2)
````

## Ejercicio 9
Concatena los siguientes arrays horizontamente

```Python
a = np.arange(10).reshape(2,-1)
b = np.repeat(1, 10).reshape(2,-1)
```

````python
a = np.arange(10).reshape(2,-1)
b = np.repeat(1, 10).reshape(2,-1)
c9_1 = np.concatenate([a, b], axis=1)
c9_2 = np.hstack([a, b])

print(a)
print()
print(b)
print()
print(c9_1)
print()
print(c9_2)
print()
````

## Ejercicio 10
Encuentra todos los elementos en común entre ambos arrays. [Para ello usa la función `intersect1d`](https://numpy.org/doc/stable/reference/generated/numpy.intersect1d.html)

```Python
a = np.array([1,2,3,2,3,4,3,4,5,6])
b = np.array([7,2,10,2,7,4,9,4,9,8])
```

````python
a = np.array([1,2,3,2,3,4,3,4,5,6])
b = np.array([7,2,10,2,7,4,9,4,9,8])

comunes_ab = np.intersect1d(a, b)
print(comunes_ab)
````

## Ejercicio 20
Dado el siguiente array

```Python
x = np.random.randint(10, 20, size = (5,2,6))
```

1. ¿Cuántos elementos tiene el array?
2. ¿Cuántas dimensiones?
3. ¿Cuánta es la longitud cada dimensión?

````python
x = np.random.randint(10, 20, size = (5,2,6))

# 1. 
print(f'Nº de elementos de x: {x.size}')

# 2. 
print(f'Nº de dimensiones de x: {x.ndim}')

# 3. 
print(f'Longitud de cada dimensión de x: {x.shape}')
````
