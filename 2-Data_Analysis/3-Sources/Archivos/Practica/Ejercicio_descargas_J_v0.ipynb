## Ejercicio descargas

Python es un lenguaje de propósito general, por lo que podemos desarrollar otros programas,
aparte de ciencia de datos, que nos permitan montar dashboards, crear APIs o automatizar
procesos. En este caso vas a automatizar un proceso, en el que tendrás que desarrollar un
script que ordene todos los archivos de una carpeta, dependiendo de la extensión de los
mismos.


El objetivo es ordenar la típica carpeta de cajón desastre donde se nos acumulan archivos de
todos los formatos como: imágenes, words, presentaciones, excels, ejecutables, zips, etc... Y
que no vamos a ir viendo archivo a archivo, ordenándolos.... La solución suele ser eliminarlo
todo. Vamos a intentar no tener que llegar a eso, ordenando los ficheros automáticamente en
carpetas y revisar las que nos interesen. La típica carpeta donde aplicaríamos este programa es
la de “Descargas”, donde se suelen acumular muchos archivos temporales.


Por tanto, **el programa tiene que guardar los archivos de la carpeta “Descargas” (o cualquier
otra donde acumules mucho archivo temporal) en los siguientes directorios dentro de la
propia carpeta “Descargas”**:
- Imagenes
- Documentos
- Software
- Otros

Cada vez que ejecutes el script, automáticamente se ordenarán todos los archivos en sus
correspondientes carpetas.


```python
doc_types = ('.doc', '.docx', '.txt', '.pdf', '.xls', '.ppt', '.xlsx', '.pptx')
img_types = ('.jpg', '.jpeg', '.png', '.svg', '.gif')
software_types = ('.exe', '.py','.ipynb')
```

```python
import os
import shutil
```

```python
midir = r'C:\Users\julio\Downloads\prueba'

os.chdir(midir)
os.getcwd()
```

```python
subdirecs = {'Imagenes': img_types, 
             'Documentos': doc_types, 
             'Software': software_types
            }
subdirec_otro = 'Otros'
lst_subdirecs = list(subdirecs.keys())
len_lst_subdirecs = len(lst_subdirecs)

elms = os.listdir(midir)

for elm in elms:
    ruta_elm = os.path.join(midir, elm)
    if os.path.isfile(ruta_elm):
        nom_f, ext = os.path.splitext(ruta_elm)
        sigue = 1
        i = 0
        while sigue == True and i in range(len_lst_subdirecs):
            if ext in subdirecs[lst_subdirecs[i]]:
                try:
                    os.mkdir(lst_subdirecs[i])
                except:
                    pass
                shutil.move(ruta_elm, f"./{lst_subdirecs[i]}/")
                sigue = False
            i += 1
            
        if sigue == True:
            try:
                os.mkdir(subdirec_otro)
            except:
                pass
            
            try:
                shutil.move(ruta_elm, f"./{subdirec_otro}/")
            except:
                # No se moverá el fichero si ya existe uno con el mismo nombre en la carpeta de destino 
                pass
```

```python
def ordenar_fichs(direc, subdirecs, subdirec_otro):
    """
    Mueve ficheros situados en un directorio a subdirectorios 
    especificados. Solo mueve los ficheros situados en el 
    directorio raíz especificado (no los situados en subdirectorios). 
    
    Entradas: 
       direc: directorio de trabajo, donde se encuentran los ficheros a 
              mover. Tipo: str. 
       subdirecs: diccionario en el que las claves son nombres de 
                  subdirectorios, y los valores son listas o tuplas con 
                  extensiones de archivo (str). Los ficheros con extensión 
                  igual a uno de los valores de la lista correspondiente a 
                  una clave subdirec se moverán a la carpeta direc/subdirec. 
                  Esa carpeta se creará si no existe. 
       subdirec_otro: nombre del subdirectorio al que Tipo: str. 

    Salidas: None

    """
    dir_act = os.getcwd()

    os.chdir(direc)

    lst_subdirecs = list(subdirecs.keys())
    len_lst_subdirecs = len(lst_subdirecs)

    elms = os.listdir(direc)

    for elm in elms:
        ruta_elm = os.path.join(direc, elm)
        if os.path.isfile(ruta_elm):
            nom_f, ext = os.path.splitext(ruta_elm)
            sigue = 1
            i = 0
            while sigue == True and i in range(len_lst_subdirecs):
                if ext in subdirecs[lst_subdirecs[i]]:
                    try:
                        os.mkdir(lst_subdirecs[i])
                    except:
                        pass
                    shutil.move(ruta_elm, f"./{lst_subdirecs[i]}/")
                    sigue = False
                i += 1
            
            if sigue == True:
                try:
                    os.mkdir(subdirec_otro)
                except:
                    pass
            
                try:
                    shutil.move(ruta_elm, f"./{subdirec_otro}/")
                except:
                    # No se moverá el fichero si ya existe uno con el mismo nombre en la carpeta de destino 
                    pass

    os.chdir(dir_act)
```

```python
ordenar_fichs(direc=midir, subdirecs=subdirecs, subdirec_otro='Otros')
```
