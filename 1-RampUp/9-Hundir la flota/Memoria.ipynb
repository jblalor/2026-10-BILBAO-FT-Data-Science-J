## Características particulares de la implementación

- Implementación en 3 ficheros .py: 
    - HLF.py: main.
    - variables.py: parámetros. 
    - funciones.py: funciones. 
- Input del usuario: 
    - Validación del input del usuario, que se recibe por teclado. 
    - Cada vez que se pide input al usuario, se da la opción de salir del programa. 
    - Celdas recibidas en un único input, en formato LetraNúmero (letra para la columna, número para la fila). 
- Modos de juego normal y demo. En el demo se muestran los barcos del tablero de la máquina durante el juego. 
- Dimensiones de tableros, barcos, y separación mínima entre barcos configurables independientemente para cada jugador. 
- Posibilidad de colocar los barcos en el tablero propio automáticamente o manualmente, uno a uno, durante la ejecución del programa. 
- La separación mínima entre barcos puede ser cualquier entero no negativo. 
    - Al hundir un barco, las celdas circundantes en un radio de la separación mínima se descubren (configurable). 
- Distintos niveles de dificultad del juego, definidos en formato 'i[+]'. 
    - La primera parte es un entero, que define el nº máximo de casillas que sondea la máquina en su turno antes de disparar. 
    - Opcionalmente, se puede incluir '+'. Si se incluye, se da cierta inteligencia a los disparos de la máquina: 
        - Para futuros disparos, se priorizan las celdas adyacentes a la celda del barco tocado y no hundido. 
        - Se trata de determinar la orientación del barco tocado. 
            - Mirando si hay otras celdas tocadas dentro del radio de separación mínima entre barcos. 
            - Mirando el barco de mayor tamaño que cabe pasando por la última celda tocada en cada dirección. 
- Durante el juego se van mostrando por pantalla los tableros y mensajes con los resultados de los disparos y otra información del desarrollo del juego, como vidas de cada jugador, disparos hechos por cada jugador, y jugadas transcurridas. También se tiene la opción de mostrar además figuras hechas con matplotlib incluyendo ambos tableros.
