## Mini Hundir la Flota
#### Primeros pasos
Vamos a empezar a programar trozos de código que nos servirán para el HLF:

* Crear un Tablero como una lista de listas de 10x10 que contenga caracteres espacio: " "
* Insertar dos barcos sustituyendo en el Tablero los espacios en blanco por el caracter "B":
    * Uno de 4 casillas en horizontal: casillas [1][0], [1][1], [1][2], [1][3]
    * Uno de 3 casillas en vertical:   casillas [3][3], [4][3], [5][3]
* Escribir una celda de código separada de las anteriores para simular que el usuario está efectuando un disparo: 
    *  Solicite al usuario dos coordenadas `i,j` mediante dos inputs
    *  Compruebe si esas coordenadas están ocupada por una casilla barco (que haya una en "B")
        * Si hay una "B":
            * Cambiar el contenido de esa posición por una "x"
            * Imprimir por pantalla el texto "Tocado en posición `i,j`"
        * Si hay un " ":
            * Cambiar el contenido de esa posición por una "o"
            * Imprimir por pantalla el texto "Agua"
    * Imprimir por pantalla el Tablero
* Para comprobar que funciona bien la celda de disparar ejecutarla varias veces metiendo distintas coordenadas

### Bonus
* En la celda de disparar, si el usuario acierta a tocar un barco, vuelve a disparar
* Resuélvelo con funciones:
    * Una para solicitar una coordenada válida a la que llamarás dos veces, una par i y otra par j
    * Una `disparar` que devuelva `True` si el usuario ha acertado o `False` si no.
    


```python
import pprint

# Creación de tablero vacío de 10x10

ncol = 10
nfil = 10

tablero = []
for i in range(nfil):
    linea = []
    for j in range(ncol):
        linea.append(" ")
    tablero.append(linea)
tablero
```

```python
# Colocación de 2 barcos en el tablero

n_barcos = 2
len_barcos = (4, 3)
ori_barcos = ('h', 'v')
xy_ini_barcos = ((1, 0), (3, 3))

for i in range(n_barcos):
    long_barco = len_barcos[i]
    x_ini = xy_ini_barcos[i][1]
    y_ini = xy_ini_barcos[i][0]
    ori = ori_barcos[i]
    for j in range(long_barco):
        if ori == 'h':
            tablero[y_ini][x_ini + j] = "B"
        elif ori == 'v':
            tablero[y_ini+j][x_ini] = "B"

pprint.pprint(tablero)
```

```python
# Disparo, volviendo a disparar hasta que se falle

lst_coords = ['x', 'y']
dims_tablero = [ncol, nfil]
xy_disp = [None, None]

acertado = True
while acertado == True:
    # Recogida de datos de entrada del usuario 
    for i in range(len(lst_coords)):
        coord_ok = False
        while not coord_ok:
            xoy_inp = input(f"Introduce la coordenada {lst_coords[i]} de tu disparo: ")
            try:
                xoy = int(xoy_inp)
                if xoy in range(dims_tablero[i]):
                    xy_disp[i] = xoy
                    coord_ok = True
                else:
                    print(f"ERROR. Dato {lst_coords[i]} introducido incorrecto. Introduce un entero entre 0 y {dims_tablero[i]-1}.")
            except Exception as e:
                print(f"ERROR. Dato {lst_coords[i]} introducido incorrecto. Introduce un entero entre 0 y {dims_tablero[i]-1}.")
                print(f"ERROR: {e}")

    # Actualización del tablero 
    if tablero[xy_disp[1]][xy_disp[0]] == 'B':
        tablero[xy_disp[1]][xy_disp[0]] = 'x'
        acertado = True
        print(f"Tocado en posición ({xy_disp[0]}, {xy_disp[1]}).")
    elif tablero[xy_disp[1]][xy_disp[0]] == ' ':
        tablero[xy_disp[1]][xy_disp[0]] = 'o'
        acertado = False
        print("Agua")

    pprint.pprint(tablero)
    print()
```

```python
def pedir_coord(msg_in, coord_min, coord_max):
    """
    Recoge coordenada como input del usuario. 
    El input del usuario esperado es un número entero entre coord_min y coord_max. 

    Entradas: 
       msg_in: Mensaje que se muestra al usuario al pedirle el dato. Tipo: str. 
       coord_min: mínimo valor admisible de la coordenada. Tipo: int. 
       coord_max: máximo valor admisible de la coordenada. Tipo: int. 
    
    Salidas: 
       xoy: coordenada introducida por el usuario. Tipo: int. 
    
    """

    coord_ok = False
    while not coord_ok:
        xoy_inp = input(msg_in)
        try:
            xoy = int(xoy_inp)
            if xoy in range(coord_min, coord_max):
                coord_ok = True
            else:
                print(f"ERROR. Dato introducido incorrecto. Introduce un entero entre {coord_min} y {coord_max}.")
        except Exception as e:
            print(f"ERROR. Dato introducido incorrecto. Introduce un entero entre {coord_min} y {coord_max}.")
            print(f"ERROR: {e}")
    return xoy


def disparar(tablero, x, y, strs):
    """
    Simula un disparo. 

    Entradas: 
       tablero: tablero con los barcos colocados. Lista de listas, en la que cada elemento 
                de cada lista anidada es un string. En las posiciones donde hay barco, el 
                string es 'B'. 
       x, y: coordenadas donde se dispara. Entero positivo, incluyendo 0. Tipo: int. 
       strs: lista de strings que puede haber en cada celda, en este orden: 
             string en las celdas con: 
             [agua (sin disparo), barco, disparo fallado (agua), disparo acertado]. Tipo: list. 
    
    Salidas: 
       acertado: booleano que vale True si el disparo acierta en una posición con barco 
                 previamente no disparada. 
       tablero: tablero actualizado con el nuevo disparo. 
    
    """

    acertado = False

    # Actualización del tablero 
    if tablero[y][x] == strs[1]:
        tablero[y][x] = strs[3]
        acertado = True
    elif tablero[y][x] == strs[0]:
        tablero[y][x] = strs[2]
        acertado = False

    return acertado, tablero
```

```python
# Resolución con funciones 

cods = [' ', 'B', 'o', 'x']

# Creación de tablero vacío de 10x10
ncol = 10
nfil = 10

tablero = []
for i in range(nfil):
    linea = []
    for j in range(ncol):
        linea.append(cods[0])
    tablero.append(linea)

pprint.pprint(tablero)
print()


# Colocación de 2 barcos en el tablero

n_barcos = 2
len_barcos = (4, 3)
ori_barcos = ('h', 'v')
xy_ini_barcos = ((1, 0), (3, 3))

for i in range(n_barcos):
    long_barco = len_barcos[i]
    x_ini = xy_ini_barcos[i][1]
    y_ini = xy_ini_barcos[i][0]
    ori = ori_barcos[i]
    for j in range(long_barco):
        if ori == 'h':
            tablero[y_ini][x_ini + j] = cods[1]
        elif ori == 'v':
            tablero[y_ini+j][x_ini] = cods[1]

pprint.pprint(tablero)
print()


# Disparo, volviendo a disparar hasta que se falle
lst_coords = ['x', 'y']
dims_tablero = [ncol, nfil]
xy_disp = [None, None]

acert = True
while acert == True:
    # Recogida de datos de entrada del usuario 
    for i in range(len(lst_coords)):
        msg_i = f"Introduce la coordenada {lst_coords[i]} de tu disparo: "
        xy_disp[i] = pedir_coord(msg_i, 0, dims_tablero[i])

    # Disparo y actualización del tablero
    print(f'Disparando a posición {tuple(xy_disp)}')
    acert, tablero = disparar(tablero, xy_disp[0], xy_disp[1], cods)

    if acert == True:
        print(f"Tocado en posición ({xy_disp[0]}, {xy_disp[1]}).")
    elif tablero[xy_disp[1]][xy_disp[0]] == cods[2]:
        print("Agua")

    pprint.pprint(tablero)
    print()
    
```
