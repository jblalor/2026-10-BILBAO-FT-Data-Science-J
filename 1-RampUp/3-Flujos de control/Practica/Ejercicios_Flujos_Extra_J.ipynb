## Ejercicio Extra
Programa una calculadora que

* Pregunte qué operación quieres realizar (multiplicación/división/suma/resta/potencia/salir)
* Pida dos números 
* Realice la operación
* Siga funcionando hasta que el usuario elija salir

#### Bonus:
* Que guarde todas las operaciones realizadas y las imprima por pantalla cuando el usuario elija esa opción
* Que sea robusta frente a entradas erróneas del usuario o frente a operaciones como dividir por 0

```python
import math
import pprint
import time
```

```python
ops = ['multiplicación', 'división', 'suma', 'resta', 'potencia', 'imprimir', 'salir']
simb_ops = ['*', '/', '+', '-', '^']

oper = 0
rsts = []
str_ops = []

while oper != 'salir':
    oper = 0
    while oper not in ops:
        oper = input(f"¿Qué operación quieres realizar (opciones disponibles: {ops})?").lower()
        if oper not in ops:
            print("El valor introducido (", oper, ") no es una de las operaciones disponibles.")
            print("Operaciones disponibles:", ops)
            print()
        else:
            print()
            print("Operación seleccionada:", oper)

    if oper in ops[:5]: 
        simb_op = simb_ops[ops.index(oper)]
        nums = []
        for i in range(2):
            numok = False
            while not numok:
                try:
                    nums.append(float(input(f"Introduce número {i+1}: ")))
                    print(f"Número {i+1}: {nums[i]}")
                    numok = True
                except Exception as e:
                    print(e)
                    print(f"Valor incorrecto para número {i+1}. Introducir un valor numérico.")
        
        match oper:
            case 'multiplicación':
                rst = nums[0] * nums[1]
            case 'división':
                try:
                    rst = nums[0] / nums[1]
                except Exception as e:
                    print(e)
                    rst = e
            case 'suma':
                rst = nums[0] + nums[1]
            case 'resta':
                rst = nums[0] - nums[1]
            case 'potencia': 
                rst = nums[0] ** nums[1]
        
        str_ops.append(f"{nums[0]} {simb_op} {nums[1]} = {rst}")
        print(str_ops[-1])
        print()

    if oper == ops[5]: 
        print("Imprimiendo todas las operaciones realizadas")
        for e in str_ops:
            print(e)

```
