### 1. Utiliza map para elevar al cuadrado cada número en la lista [1, 2, 3, 4, 5].

Solución esperada: [1, 4, 9, 16, 25]

```python
lst = [1, 2, 3, 4, 5]
cuadrados = list(map(lambda x: x**2, lst))
print(cuadrados)
```

### 2. Filtrar números pares de una lista:

Emplea filter para obtener sólo los números pares de la lista [1, 2, 3, 4, 5, 6].

Solución esperada: [2, 4, 6]

```python
lst2 = [1, 2, 3, 4, 5, 6]
lst2_filtr = list(filter(lambda x: x % 2 < 0.1, lst2))
print(lst2_filtr)
```

### 3. Convertir grados Celsius a Fahrenheit:

Usa map y una función lambda para convertir una lista de temperaturas en grados Celsius [0, 10, 20, 30] a Fahrenheit.

Solución esperada: [32.0, 50.0, 68.0, 86.0]

```python
lst_C = [0, 10, 20, 30]
lst_F = list(map(lambda x: 9/5*x + 32, lst_C))
print(lst_F)
```

### 4. Sumar los cuadrados de los números pares de una lista:

Combina filter, map y reduce (importa reduce desde functools) para sumar los cuadrados de los números pares en [1, 2, 3, 4, 5, 6].

Solución esperada: 56

```python
lst4 = [1, 2, 3, 4, 5, 6]

pares4 = filter(lambda x: x % 2 < 0.1, lst4)
lst_pares4 = list(pares4)
print(lst_pares4)

# cuadrados = map(lambda x: x**2, list(pares))
cuadrados4 = map(lambda x: x**2, lst_pares4)
lst_cuadrados4 = list(cuadrados4)
print(lst_cuadrados4)

from functools import reduce
suma_cuadrados = reduce(lambda x,y: x+y, lst_cuadrados4)
print(suma_cuadrados)
```

### 5. Encadenar map, filter, y reduce para resolver un problema complejo:

Dada una lista de strings que representan números enteros y "NA" (por ejemplo, ["2", "5", "NA", "9", "10"]), convierte los números a enteros, filtra los "NA", suma el cuadrado de los números que son mayores que 5 y menores que 10.

Solución esperada: Un único valor que es la suma de los cuadrados de los números válidos.

```python
lst5 = ['2', '5', 'NA', '9', '10', '6']
# lst5 = ['2', '5', 'NA', '9', '10']

def pasar_a_int(str_in):
    rst = str_in
    try:
        rst = int(str_in)
    except:
        pass
    return rst

# Se pasan a int los números
lst5_int = list(map(pasar_a_int, lst5))
print(lst5_int)

# Se filtran los "NA"
# lst5_int_filtr = list(filter(lambda x: type(x)==int, lst5_int))
lst5_int_filtr = list(filter(lambda x: x!="NA", lst5_int))
print(lst5_int_filtr)

# Se filtran los nºs mayores que 5 y menores que 10
lst5_int_filtr_5a10 = list(filter(lambda x: x>5 and x<10, lst5_int_filtr))
print(lst5_int_filtr_5a10)

# Se calculan los cuadrados de la lista anterior 
lst_cuadrados5 = list(map(lambda x: x**2, lst5_int_filtr_5a10))
print(lst_cuadrados5)

# Se suman los cuadrados de la lista anterior
cuadrados5 = reduce(lambda x,y: x + y, lst_cuadrados5, 0)
print(cuadrados5)
```
