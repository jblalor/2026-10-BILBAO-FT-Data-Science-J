## Ejercicio 1
Somos una empresa que tiene varias tiendas de electrodomésticos. Necesitamos un programa para manejar las tiendas, ver las ventas que han tenido, cuántos empleados hay, etc... Para ello vamos a modelizar la tienda en una clase, que tendrá las siguientes características:
* Nombre clase: "Tienda"
* Atributos comunes:
    * Tipo: "Electrodomésticos"
    * Abierta: True
* Atributos propios de cada tienda:
    * Nombre - String
    * Dirección - String
    * Número de empleados - int
    * Ventas últimos 3 meses - Lista de 3 numeros
* Método para calcular las ventas de todos los meses, que devuelva un número con todas las ventas.
* Método que calcula la media de ventas de los últimos meses, por empleado
* Método que devuelve en un string el nombre de la tienda, junto con su dirección.
* Método que obtiene las ventas del último mes.
* Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

Se pide:
1. Implementar la clase "Tienda"
2. Crear tres tiendas con datos inventados
3. Comprueba en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...)
3. Calcular las ventas del último mes de todas las tiendas. Para ello usa el bucle `for`
4. Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"
5. Seguro que ya has hecho este apartado... Documenta la clase :)

```python
# 1. 
class Tienda:
    """
    Clase de las tiendas de nuestra empresa. 
    Parámetros de inicialización:
       - nombre: nombre de la tienda. Tipo: str. 
       - direccion: dirección de la tienda. Tipo: str. 
       - n_emp: nº de empleados. Tipo: int. 
       - ventas_ult_3m: Ventas de los últimos 3 meses. 
                        Tipo: list de 3 floats. El último elemento
                        corresponde al último mes. 
    Otros parámetros: 
       - tipo: tipo de tienda. 
       - abierta. Tipo: bool. 
    """
    tipo = "Electrodomésticos"
    abierta = True

    def __init__(self, nom:str, direc:str, n_emp:int, vnts_ult_3m:list):
        self.nombre = nom
        self.direccion = direc
        self.num_empleados = n_emp
        self.ventas_ult_3m = vnts_ult_3m

    
    def suma_ventas(self):
        """
        Calcula la suma de ventas de los últimos 3 meses. 
        """
        return sum(self.ventas_ult_3m)

    
    def ventas_x_empl(self):
        """
        Calcula la media de ventas de los últimos 3 meses por empleado. 
        """
        return self.suma_ventas() / self.num_empleados

    
    def nombre_y_direccion(self):
        """
        Devuelve en un string el nombre y dirección de la tienda. 
        """
        return f'Tienda: {self.nombre}. Dirección: {self.direccion}'


    def ventas_ultimo_mes(self):
        """
        Devuelve las ventas del último mes. 
        """
        return self.ventas_ult_3m[-1]

    
    def proyec_ventas(self, inv_marketing):
        """
        Calcula la proyección de ventas de los últimos meses en caso de 
        haber invertido en marketing. 
        Entradas: 
           inv_marketing: dinero invertido en marketing en la tienda. 
        Salidas:
           Vector con las ventas proyectadas de los últimos 3 meses. 
           También se reescribe el atributo ventas_ult_3m con esta 
           proyección. 
        """
        if inv_marketing < 1000:
            factor = 1.2
        else:
            factor = 1.5
        
        ventas_proy = [ventas * factor for ventas in self.ventas_ult_3m]
        self.ventas_ult_3m = ventas_proy

        return ventas_proy


# 2. 
tienda1 = Tienda('ElectroFlash Avda', 'Avenida Diagonal, 16', 3, [23., 19., 26.])
inv_mrk_1 = 800.
tienda2 = Tienda('ElectroFlash Zazpi', 'Calle Mayor, 8', 2, [12., 14., 11.])
tienda2.abierta = False
inv_mrk_2 = 1200.
tienda3 = Tienda('Hiper ElectroFlash', 'Paseo de la Castellana, 45', 5, [47., 39., 45.])
inv_mrk_3 = 1500.

tiendas = [tienda1, tienda2, tienda3]
v_inv_mrk = [800., 1200., 1500.]

# 3. 
for i, tienda in enumerate(tiendas):
    print(f'Tienda {tienda.nombre}: ')
    print(tienda.__dict__)
    print('Tipo de tienda:', tienda.tipo)
    print('Abierta:', tienda.abierta)
    print('Nombre:', tienda.nombre)
    print('Dirección: ', tienda.direccion)
    print('Número de empleados: ', tienda.num_empleados)
    print('Ventas de los últimos 3 meses: ', tienda.ventas_ult_3m)
    print('Suma de ventas de los últimos 3 meses: ', tienda.suma_ventas())
    print('Suma de ventas de los últimos 3 meses por empleado: ', tienda.ventas_x_empl())
    print('Nombre y dirección: ', tienda.nombre_y_direccion())
    print('Ventas del último mes: ', tienda.ventas_ultimo_mes())
    print(f'Suma de ventas proyectadas de los últimos 3 meses en caso de haber invertido {v_inv_mrk[i]} euros en marketing: {tienda.proyec_ventas(v_inv_mrk[i])}')
    print('Ventas de los últimos 3 meses (actualizadas): ', tienda.ventas_ult_3m)
    print('Suma de ventas de los últimos 3 meses (actualizada): ', tienda.suma_ventas())
    print('Suma de ventas de los últimos 3 meses por empleado (actualizada): ', tienda.ventas_x_empl())
    print('Ventas del último mes (actualizadas): ', tienda.ventas_ultimo_mes())
    print()
    print()

# 4.
ventas_ult_mes = []
for tienda in tiendas:
    ventas_ult_mes.append(tienda.ventas_ultimo_mes())
print(ventas_ult_mes, sum(ventas_ult_mes))
print()

# 5. 
for tienda in tiendas:
    if "Avenida" in tienda.direccion:
        print(tienda.nombre)
```

## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
# 1. 
class Perro:
    """
    Clase de mascotas caninas. 
    Parámetros de inicialización: 
       - raza. Tipo: str. 
       - col_pelo: color de su pelo. Tipo: str. Valor por defecto: Marrón. 
       - duenio: dueño. Tipo: str. Valor por defecto: None. 
    Otros parámetros: 
       - patas: nº de patas. Tipo: int. Valor por defecto: 4. 
       - orejas: nº de orejas. Tipo: int. Valor por defecto: 2. 
       - ojos: nº de ojos. Tipo: int. Valor por defecto: 2. 
       - vel: velocidad. Tipo: float. Valor por defecto: 0. [m/s] 
    """
    
    patas = 4
    orejas = 2
    ojos = 2
    vel = 0.

    def __init__(self, raza:str, c_pelo:str="Marrón", duenio:str=None):
        self.raza = raza
        self.col_pelo = c_pelo
        self.duenio = duenio


    def andar(self, aumento_velocidad):
        """
        Aumenta la velocidad (atributo .vel) del perro. 
        Entradas: 
           - aumento_velocidad: aumento de velocidad. Tipo: float [m/s]
        """

        self.vel += aumento_velocidad


    def parar(self):
        """
        Para al perro, cambiando su atributo vel a 0.  
        """

        self.vel = 0.


    def ladrar(self, msg):
        """
        Ladrido compuesto por "GUAU!" y un mensaje adicional. 
        Entradas: 
           - msg: Mensaje incluido en el ladrido. Tipo: str. 
        Salidas: 
           - Concatenación de "GUAU!" y msg. Tipo: str. 
        """
        return "GUAU!" + " " + msg

# 2. 
perro1 = Perro('carlino', 'gris')

# 3. 
print(perro1.__dict__)
print(perro1.patas)
print(perro1.orejas)
print(perro1.ojos)
print(perro1.vel)
print(perro1.raza)
print(perro1.col_pelo)
print(perro1.duenio)
print()

# 4. 
perro1.andar(5.)
print(perro1.vel)
perro1.andar(7.)
print(perro1.vel)
print()

# 5. 
perro1.parar()
print(perro1.vel)
print()

ladrido1 = perro1.ladrar('¡Quiero pasear!')
print(ladrido1)
```
