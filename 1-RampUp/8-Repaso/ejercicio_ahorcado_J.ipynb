## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
import os
import time

def ahorcado(palabra, vidas):
    """
    Simulación del juego del ahorcado. 

    Entradas: 
       palabra: string. Las letras con diacríticos se deberían introducir sin ellos 
                tanto en esta variable como en el input por teclado que pide la 
                función, ya que se interpretan como letras distintas en esta función. 
       vidas: nº de vidas. Entero mayor que 0. 

    Salidas: None

    """

    palabra = palabra.upper()
    descub = False
    errores = 0
    l_palabra = len(palabra)
    palabra_ = '_' * l_palabra
    lst_palabra_ = list(palabra_)
    str_show = ' '.join(palabra_)
    print(str_show)
    print(f'Te quedan {vidas} vidas.')
    print()
    print()

    while vidas > 0.1 and descub == False:
        os.system('cls')
        esletra = False
        while not esletra:
            letra = input('Inserta una letra: ')
            print(f'Dato introducido: {letra}')
            if not (letra.isalpha() and len(letra) == 1):
                print('Entrada incorrecta. Inserta una única letra.')
            else:
                esletra = True
        
        letra = letra.upper()

        if letra in palabra and not letra in lst_palabra_:
            indices = [i for i, l in enumerate(palabra) if l == letra]
            for i in indices:
                lst_palabra_[i] = letra
            str_show = ' '.join(lst_palabra_)
            print('¡La palabra contiene esa letra!')
            if '_' not in lst_palabra_:
                descub = True
        else:
            errores += 1
            vidas -= 1
            if letra in lst_palabra_:
                print('¡Esa letra ya se había descubierto!')
            else:
                print('La palabra no contiene esa letra.')
        print()
        print(str_show)
        print()
        if descub == True:
            print('¡Palabra descubierta!')
        else:
            print(f'Te quedan {vidas} vidas.')
            if errores > 0.1:
                print(f'Llevas {errores} errores.')
        print()
        print()
        if vidas < 0.1:
            print('Fin del juego.')

    return None


vds = 6

palabra_ej = 'supercalifragilisticoespialidoso'

ahorcado(palabra_ej, vds)
````
