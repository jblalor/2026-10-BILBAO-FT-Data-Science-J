## Ejercicio 4
Tenemos las siguientes variables:

```Python
A = 4
B = "Text"
C = 4.1
```

Comprueba:
1. Si A y B son equivalentes
2. Si A y C NO son equivalentes
3. Si A es mayor que C 
4. Si C es menor o igual que A
5. Si B NO es equivalente a C

````python
if A == B:
    print("Son iguales")
else:
    print("No lo son")
````

````python
A = 4
B = "Text"
C = 4.1

# Comprobación 1
if A == B:
    txt11 = " "
    txt12 = " ="
else:
    txt11 = " no "
    txt12 = ";"
print(f"1. A y B{txt11}son iguales. A = {A}{txt12} B = {B}.")

# Comprobación 2
if not (A == C):
    txt21 = " no "
    txt22 = ";"
else:
    txt21 = " "
    txt22 = " ="
print(f"2. A y C{txt21}son iguales. A = {A}{txt22} C = {C}.")

# Comprobación 3
if A>C:
    print(f"3. A es mayor que C (A = {A} > C = {C}).")
else:
    print(f"3. A no es mayor que C (A = {A}; C = {C}).")

# Comprobación 4
if C<=A:
    print(f"4. C es menor o igual que A (C = {C} <= A = {A}).")
else:
    print(f"4. C no es menor o igual que A (C = {C}; A = {A}).")

# Comprobación 5
if not (B == C):
    print(f"5. B no es equivalente a C. B = {B}; C = {C}.")
else:
    print(f"5. B es equivalente a C. A = {A} = B = {B}.")
````

## Ejercicio 5
Crea un programa donde se recojan dos inputs del usuario, y el output del programa sea si esos inputs son iguales o no

````python
inps = []

for i in range(2):
    inps.append(input(f"Introduce la entrada  {i+1}:"))

if inps[0] == inps[1]:
    txt = ""
else:
    txt = "no"

print(f"La primera entrada introducida ({inps[0]}) {txt} es equivalente a la segunda ({inps[1]}).")
````

## Ejercicio 7
Crea un programa que recoja dos inputs. Tiene que comprobar si su suma es igual, superior o inferior a 10, cada condición por separado

````python
inp1 = input("Introducir primer valor:")
inp2 = input("Introducir segundo valor:")

suma = float(inp1) + float(inp2)

if suma == 10:
    txt = "igual a 10."
elif suma < 10:
    txt = f"inferior a 10 ({suma})."
else:
    txt = f"superior a 10 ({suma})."

print(f"La suma de los dos valores introducidos ({inp1} y {inp2}) es {txt}")
````

## Ejercicio 9
Para este ejercicio vamos a poner en práctica [las funciones built in](https://docs.python.org/3/library/functions.html).

2. Suma todos los elementos de la lista anterior
3. Redondea este float a 3 dígitos decimales: 63.451256965

## Ejercicio 10
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "Lo primero de todo, cómo están los máquinas"

1. Pasarlo todo a mayúsculas
2. Pasarlo todo a minúsculas
3. Pasar a mayúsculas solo las iniciales de las palabras
4. Crea una lista dividiendolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

````python
str_inp = "Lo primero de todo, cómo están los máquinas"

str_mayusc = str_inp.upper()
print(f"1. {str_mayusc}")

str_minusc = str_inp.lower()
print(f"2. {str_minusc}")

str_3 = str_inp.title()
print(f"3. {str_3}")

lst_str = str_inp.split()
print(f"4. {lst_str}")

str5 = str_inp.replace(",", ";")
print(f"5. {str5}")

set_elim = set(["a", "á"])
str6 = ''.join([c for c in str_inp if c not in set_elim])
print(f"6. {str6}")
````
