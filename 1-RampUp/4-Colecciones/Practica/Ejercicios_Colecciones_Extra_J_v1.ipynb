```python
import pprint

opers = ['vt', 'b', 'a', 'e', 'alq', 'dev']
opers_ = ['visualizar todos (vt)', 'buscar (b)', 'añadir (a)', 'eliminar (e)', 'alquilar (alq)', 'devolver (dev)']
# libros_aux = {libro["Titulo"]: {k: v for k, v in libros.items() if k != "Titulo"} for libro in libros}
lst_str_old = ['á', 'é', 'í', 'ó', 'ú', 'à', 'è', 'ù', 'â', 'ê', 'î', 'ô', 'û', 'ë', 'ï', 'ü', '-', '_', ':', ',', '.']
lst_str_new = ['a', 'e', 'i', 'o', 'u', 'a', 'e', 'u', 'a', 'e', 'i', 'o', 'u', 'e', 'i', 'u', ' ', ' ', '', '', '']
d_cambio_str = dict(zip(lst_str_old, lst_str_new))

libros_aux = {libro["Titulo"]
              .lower()
              .translate(str.maketrans(d_cambio_str))
              .strip(): 
              libro for libro in libros}

titulos = [libro["Titulo"] for libro in libros]
titulos_aux = list(libros_aux.keys())
# titulos_aux = [titulo.lower().translate(str.maketrans(d_cambio_str)).strip() for titulo in titulos]

ttl = None

op = 0
while op not in opers:
    op = input(f"¿Qué operación quieres realizar (opciones disponibles: {opers_})?").lower()
    if op not in opers:
        print(f"El valor introducido ({op}) no es una de las operaciones disponibles.")
        print("Operaciones disponibles:", opers_)
        print("(Introducir solo el texto entre paréntesis)")
        print()
    else:
        op_slc = opers_[opers.index(op)]
        print(f"Operación seleccionada: {op_slc}")
        print()

if op == 'vt': 
    print("Se imprimen a continuación los datos de todos los libros existentes en la biblioteca: ")
    print()
    pprint.pprint(libros)
elif op == 'a':
    ttl_inp = input(f"Introducir título del libro a añadir: ")
    print(f"Título introducido: {ttl_inp}")
    ttl_inp_aux = ttl_inp.lower().translate(str.maketrans(d_cambio_str)).strip()
    if ttl_inp in titulos or ttl_inp_aux in titulos_aux:
        print('ERROR: libro existente con el mismo título. No se añade ningún libro.')
        print('Datos del libro existente:')
        pprint.pprint(libros_aux[ttl_inp_aux])
    else:
        autor = input("Introducir autor del libro a añadir: ")
        print(f"Autor introducido: {autor}")
        libro_n = {'Titulo': ttl_inp, 'Autor': autor, 'Alquilado': False}
        libros.append(libro_n)
        print("Libro añadido a lista de libros existentes en la biblioteca:")
        print(libros[-1])
elif op in ['b', 'e', 'alq', 'dev']:
    ttl_inp = input(f"Introducir título del libro existente a {op_slc}: ")
    print(f"Título introducido: {ttl_inp}")
    ttl_inp = ttl_inp.lower().translate(str.maketrans(d_cambio_str)).strip()
    ttl = []
    encontrado = False
    try:
        # Se busca título con coincidencia "exacta"
        ttl = libros_aux[ttl_inp]['Titulo']
        print(f"Libro encontrado con título: {ttl}")
        encontrado = True
    except:
        # Se busca(n) título(s) que contenga(n) el título introducido
        print("No encontrado libro con coincidencia exacta de título introducido.")
        print("Buscando título introducido como substring dentro de títulos existentes...")
        ttl = [libros_aux[k]['Titulo'] for k in libros_aux.keys() if ttl_inp in k]
        ttl_aux = [e.lower().translate(str.maketrans(d_cambio_str)).strip() for e in ttl]
        if len(ttl) < 0.1:
            print("ERROR: no encontrado ningún libro con el título indicado.")
        else: 
            print(f"Encontrada(s) la(s) siguiente(s) {len(ttl)} posible(s) coincidencia(s):")
            for i, e in enumerate(ttl_aux):
                print(i, libros_aux[e])
            print()
            ittl_inp = -1
            ittl_inp = input(f"Introducir el índice del libro a {op_slc} de la lista anterior \n(introducir un entero de 0 a {len(ttl)-1}; si no se introduce un valor válido \nse asumirá que ninguno de los libros listados es el buscado).")
            ittl_inp = int(ittl_inp)
            print(f"Índice del libro introducido: {ittl_inp}")
            if ittl_inp in range(len(ttl)):
                ttl = ttl[ittl_inp]
                ttl_aux = ttl_aux[ittl_inp]
                print("Mostrando todos los campos del libro seleccionado: ")
                pprint.pprint(libros_aux[ttl_aux])
                encontrado = True
            else:
                print("Índice introducido fuera de rango. Ninguno de los libros indicados es el buscado.")
                
    if encontrado:
        ind = titulos.index(ttl)
        match op:
            case 'b':
                print("Mostrando todos los campos del libro buscado:")
                pprint.pprint(libros[ind])
                pass
            case 'e':
                libros.pop(ind)
                print("Eliminado libro indicado de la lista de libros de la biblioteca.")
            case 'alq':
                if libros[ind]['Alquilado'] == True:
                    print("ATENCIÓN: Libro ya alquilado.")
                else:
                    libros[ind]['Alquilado'] = True
                    print("Marcado como alquilado el libro indicado.")
            case 'dev':
                if libros[ind]['Alquilado'] == False:
                    print("ATENCIÓN: libro ya marcado como devuelto. Posible error de inventario.")
                else:
                    libros[ind]['Alquilado'] = False
                    print("Marcado como devuelto el libro indicado.")

# pprint.pprint(libros_aux)
```
